```python
# ==========================================
# DAY 2 - STEP 1
# LOAD CLEANED DATASET
# ==========================================

import pandas as pd

cleaned_path = r"D:\APL_Logistics_Project\03_Cleaned_Data\APL_Logistics_Cleaned.csv"

df = pd.read_csv(
    cleaned_path,
    encoding="utf-8-sig"
)

print("✅ Cleaned dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
# ==========================================
# DAY 2 - STEP 2
# CREATE DELAY GAP
# ==========================================

df["Delay Gap"] = (
    df["Days for shipping (real)"]
    - df["Days for shipment (scheduled)"]
)

print("✅ Delay Gap feature created!")

print("\nDelay Gap Summary:")
display(
    df["Delay Gap"].describe()
)

print("\nUnique Delay Gap Values:")
print(
    sorted(df["Delay Gap"].unique())
)
```

```python
# ==========================================
# DAY 2 - STEP 3
# DELIVERY PERFORMANCE CLASSIFICATION
# ==========================================

df["Delivery Performance"] = "On-Time"

# Canceled shipments
df.loc[
    df["Delivery Status"] == "Shipping canceled",
    "Delivery Performance"
] = "Canceled"

# Early shipments
df.loc[
    (df["Delay Gap"] < 0) &
    (df["Delivery Status"] != "Shipping canceled"),
    "Delivery Performance"
] = "Early"

# Delayed shipments
df.loc[
    (df["Delay Gap"] > 0) &
    (df["Delivery Status"] != "Shipping canceled"),
    "Delivery Performance"
] = "Delayed"

# Check results
print("✅ Delivery Performance classification created!")

print("\nDelivery Performance Counts:")
print(df["Delivery Performance"].value_counts())

print("\nDelivery Performance Percentages:")
print(
    df["Delivery Performance"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ==========================================
# DAY 2 - STEP 4
# VALIDATE DELIVERY PERFORMANCE
# ==========================================

validation_table = pd.crosstab(
    df["Delivery Status"],
    df["Delivery Performance"]
)

print("✅ Validation table created!")

display(validation_table)
```

```python
print("\nOriginal Delivery Status:")
print(df["Delivery Status"].value_counts())

print("\nEngineered Delivery Performance:")
print(df["Delivery Performance"].value_counts())
```

```python
# ==========================================
# DAY 2 - STEP 5
# OVERALL DELIVERY PERFORMANCE KPIs
# ==========================================

# Exclude canceled shipments for delivery-performance KPIs
non_canceled = df[
    df["Delivery Performance"] != "Canceled"
]

# Total non-canceled shipments
total_non_canceled = len(non_canceled)

# Shipment counts
delayed_count = (
    non_canceled["Delivery Performance"] == "Delayed"
).sum()

on_time_count = (
    non_canceled["Delivery Performance"] == "On-Time"
).sum()

early_count = (
    non_canceled["Delivery Performance"] == "Early"
).sum()

canceled_count = (
    df["Delivery Performance"] == "Canceled"
).sum()

# KPI calculations
on_time_rate = (
    on_time_count / total_non_canceled
) * 100

early_rate = (
    early_count / total_non_canceled
) * 100

delay_rate = (
    delayed_count / total_non_canceled
) * 100

cancellation_rate = (
    canceled_count / len(df)
) * 100

# Average delay across non-canceled shipments
average_delay_gap = (
    non_canceled["Delay Gap"].mean()
)

# Average delay only among delayed shipments
average_delay_delayed = (
    non_canceled.loc[
        non_canceled["Delivery Performance"] == "Delayed",
        "Delay Gap"
    ].mean()
)

print("==========================================")
print("OVERALL DELIVERY PERFORMANCE KPIs")
print("==========================================")

print(f"Total Shipments          : {len(df):,}")
print(f"Non-Canceled Shipments   : {total_non_canceled:,}")
print(f"Delayed Shipments        : {delayed_count:,}")
print(f"On-Time Shipments        : {on_time_count:,}")
print(f"Early Shipments          : {early_count:,}")
print(f"Canceled Shipments       : {canceled_count:,}")

print("\n--- KPI RESULTS ---")

print(f"On-Time Delivery Rate    : {on_time_rate:.2f}%")
print(f"Delay Rate               : {delay_rate:.2f}%")
print(f"Early Delivery Rate      : {early_rate:.2f}%")
print(f"Cancellation Rate        : {cancellation_rate:.2f}%")

print(f"\nAverage Delay Gap        : {average_delay_gap:.2f} days")
print(
    f"Avg Delay When Delayed   : "
    f"{average_delay_delayed:.2f} days"
)
```

```python
# ==========================================
# DAY 2 - STEP 6
# LATE DELIVERY RISK ANALYSIS
# ==========================================

print("==========================================")
print("LATE DELIVERY RISK ANALYSIS")
print("==========================================")

# Risk distribution
risk_counts = df["Late_delivery_risk"].value_counts()

print("\nLate Delivery Risk Counts:")
print(risk_counts)

print("\nLate Delivery Risk Percentages:")
print(
    df["Late_delivery_risk"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

# Compare risk flag with actual delivery performance
risk_vs_performance = pd.crosstab(
    df["Late_delivery_risk"],
    df["Delivery Performance"]
)

print("\nRisk Flag vs Delivery Performance:")
display(risk_vs_performance)

# Compare risk flag with original Delivery Status
risk_vs_status = pd.crosstab(
    df["Late_delivery_risk"],
    df["Delivery Status"]
)

print("\nRisk Flag vs Original Delivery Status:")
display(risk_vs_status)
```

```python
# ==========================================
# DAY 3 - STEP 1
# SHIPPING MODE DISTRIBUTION
# ==========================================

print("==========================================")
print("SHIPPING MODE ANALYSIS")
print("==========================================")

shipping_mode_counts = (
    analysis_df["Shipping Mode"]
    .value_counts()
)

print("\nShipping Mode Counts:")
print(shipping_mode_counts)

print("\nShipping Mode Percentages:")
print(
    analysis_df["Shipping Mode"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ==========================================
# DAY 3 - STEP 2
# SHIPPING MODE DELAY PERFORMANCE
# ==========================================

shipping_mode_analysis = (
    analysis_df
    .groupby("Shipping Mode")
    .agg(
        Total_Shipments=("Shipping Mode", "size"),
        Delayed_Shipments=(
            "Delivery Performance",
            lambda x: (x == "Delayed").sum()
        ),
        On_Time_Shipments=(
            "Delivery Performance",
            lambda x: (x == "On-Time").sum()
        ),
        Early_Shipments=(
            "Delivery Performance",
            lambda x: (x == "Early").sum()
        ),
        Canceled_Shipments=(
            "Delivery Performance",
            lambda x: (x == "Canceled").sum()
        ),
        Average_Delay_Gap=("Delay Gap", "mean")
    )
    .reset_index()
)

# Calculate delay rate
shipping_mode_analysis["Delay_Rate_%"] = (
    shipping_mode_analysis["Delayed_Shipments"]
    / shipping_mode_analysis["Total_Shipments"]
    * 100
)

# Calculate on-time rate
shipping_mode_analysis["On_Time_Rate_%"] = (
    shipping_mode_analysis["On_Time_Shipments"]
    /
    (
        shipping_mode_analysis["Total_Shipments"]
        - shipping_mode_analysis["Canceled_Shipments"]
    )
    * 100
)

# Sort by highest delay rate
shipping_mode_analysis = (
    shipping_mode_analysis
    .sort_values("Delay_Rate_%", ascending=False)
)

print("==========================================")
print("SHIPPING MODE DELAY PERFORMANCE")
print("==========================================")

display(
    shipping_mode_analysis.round(2)
)
```

```python
# ==========================================
# DAY 3 - STEP 3
# AVERAGE DELAY WHEN ACTUALLY DELAYED
# ==========================================

shipping_mode_delay = (
    analysis_df[
        analysis_df["Delivery Performance"] == "Delayed"
    ]
    .groupby("Shipping Mode")
    .agg(
        Delayed_Shipments=("Delay Gap", "count"),
        Average_Delay_When_Delayed=("Delay Gap", "mean"),
        Maximum_Delay=("Delay Gap", "max")
    )
    .reset_index()
)

shipping_mode_delay = (
    shipping_mode_delay
    .sort_values(
        "Average_Delay_When_Delayed",
        ascending=False
    )
)

print("==========================================")
print("AVERAGE DELAY WHEN ACTUALLY DELAYED")
print("==========================================")

display(
    shipping_mode_delay.round(2)
)
```

```python
# ==========================================
# DAY 3 - STEP 4
# SHIPPING MODE EFFICIENCY INDEX
# ==========================================

shipping_efficiency = (
    analysis_df
    .groupby("Shipping Mode")
    .agg(
        Total_Shipments=("Shipping Mode", "size"),
        Delayed_Shipments=(
            "Delivery Performance",
            lambda x: (x == "Delayed").sum()
        ),
        On_Time_Shipments=(
            "Delivery Performance",
            lambda x: (x == "On-Time").sum()
        ),
        Early_Shipments=(
            "Delivery Performance",
            lambda x: (x == "Early").sum()
        ),
        Canceled_Shipments=(
            "Delivery Performance",
            lambda x: (x == "Canceled").sum()
        )
    )
    .reset_index()
)

# Non-canceled shipments
shipping_efficiency["Non_Canceled"] = (
    shipping_efficiency["Total_Shipments"]
    - shipping_efficiency["Canceled_Shipments"]
)

# Delay rate
shipping_efficiency["Delay_Rate_%"] = (
    shipping_efficiency["Delayed_Shipments"]
    / shipping_efficiency["Non_Canceled"]
    * 100
)

# On-time rate
shipping_efficiency["On_Time_Rate_%"] = (
    shipping_efficiency["On_Time_Shipments"]
    / shipping_efficiency["Non_Canceled"]
    * 100
)

# Efficiency Index
shipping_efficiency["Efficiency_Index"] = (
    shipping_efficiency["On_Time_Rate_%"]
    - shipping_efficiency["Delay_Rate_%"]
)

# Cancellation rate
shipping_efficiency["Cancellation_Rate_%"] = (
    shipping_efficiency["Canceled_Shipments"]
    / shipping_efficiency["Total_Shipments"]
    * 100
)

# Sort best to worst
shipping_efficiency = (
    shipping_efficiency
    .sort_values("Efficiency_Index", ascending=False)
)

print("==========================================")
print("SHIPPING MODE EFFICIENCY INDEX")
print("==========================================")

display(
    shipping_efficiency.round(2)
)
```

```python
# ==========================================
# DAY 5 - STEP 3
# DELIVERY STATUS × SHIPPING MODE
# ==========================================

status_mode = (
    pd.crosstab(
        df["Shipping Mode"],
        df["Delivery Status"]
    )
    .reset_index()
)

# Ensure all expected columns exist
expected_statuses = [
    "Late delivery",
    "Advance shipping",
    "Shipping on time",
    "Shipping canceled"
]

for status in expected_statuses:
    if status not in status_mode.columns:
        status_mode[status] = 0

# Rename columns
status_mode = status_mode.rename(
    columns={
        "Late delivery": "Late_Delivery",
        "Advance shipping": "Advance_Shipping",
        "Shipping on time": "Shipping_On_Time",
        "Shipping canceled": "Shipping_Canceled"
    }
)

# Total shipments
status_mode["Total_Shipments"] = (
    status_mode["Late_Delivery"]
    + status_mode["Advance_Shipping"]
    + status_mode["Shipping_On_Time"]
    + status_mode["Shipping_Canceled"]
)

# Calculate percentages
status_mode["Late_%"] = (
    status_mode["Late_Delivery"]
    / status_mode["Total_Shipments"]
    * 100
)

status_mode["Advance_%"] = (
    status_mode["Advance_Shipping"]
    / status_mode["Total_Shipments"]
    * 100
)

status_mode["On_Time_%"] = (
    status_mode["Shipping_On_Time"]
    / status_mode["Total_Shipments"]
    * 100
)

status_mode["Canceled_%"] = (
    status_mode["Shipping_Canceled"]
    / status_mode["Total_Shipments"]
    * 100
)

# Round percentages
percentage_columns = [
    "Late_%",
    "Advance_%",
    "On_Time_%",
    "Canceled_%"
]

status_mode[percentage_columns] = (
    status_mode[percentage_columns].round(2)
)

print("==========================================")
print("DELIVERY STATUS × SHIPPING MODE")
print("==========================================")

display(status_mode)
```

```python
# ==========================================
# DAY 5 - STEP 4
# DELAY SEVERITY DISTRIBUTION
# ==========================================

delay_severity = (
    df[df["Delay Gap"] > 0]
    .groupby("Delay Gap")
    .size()
    .reset_index(name="Delayed_Shipments")
)

# Total delayed shipments
total_delayed = delay_severity["Delayed_Shipments"].sum()

# Percentage of all shipments
delay_severity["Share_of_All_Shipments_%"] = (
    delay_severity["Delayed_Shipments"]
    / len(df)
    * 100
)

# Percentage of delayed shipments
delay_severity["Share_of_Delayed_Shipments_%"] = (
    delay_severity["Delayed_Shipments"]
    / total_delayed
    * 100
)

# Severity label
delay_severity["Severity"] = (
    delay_severity["Delay Gap"]
    .map({
        1: "Minor Delay (+1 day)",
        2: "Moderate Delay (+2 days)",
        3: "Severe Delay (+3 days)",
        4: "Severe Delay (+4 days)"
    })
)

# Round
delay_severity[
    [
        "Share_of_All_Shipments_%",
        "Share_of_Delayed_Shipments_%"
    ]
] = (
    delay_severity[
        [
            "Share_of_All_Shipments_%",
            "Share_of_Delayed_Shipments_%"
        ]
    ].round(2)
)

print("==========================================")
print("DELAY SEVERITY DISTRIBUTION")
print("==========================================")

display(delay_severity)
```

```python
# ==========================================
# DAY 5 - STEP 5
# PRODUCT / CATEGORY ANALYSIS
# ==========================================

category_analysis = (
    df.groupby("Category Name")
    .agg(
        Total_Shipments=("Delivery Performance", "size"),
        Delayed=("Delivery Performance", lambda x: (x == "Delayed").sum()),
        On_Time=("Delivery Performance", lambda x: (x == "On-Time").sum()),
        Early=("Delivery Performance", lambda x: (x == "Early").sum()),
        Canceled=("Delivery Performance", lambda x: (x == "Canceled").sum()),
        Avg_Delay_Gap=("Delay Gap", "mean")
    )
    .reset_index()
)

# Non-canceled shipments
category_analysis["Non_Canceled"] = (
    category_analysis["Total_Shipments"]
    - category_analysis["Canceled"]
)

# Delay rate
category_analysis["Delay_Rate_%"] = (
    category_analysis["Delayed"]
    / category_analysis["Non_Canceled"]
    * 100
)

# On-time rate
category_analysis["On_Time_Rate_%"] = (
    category_analysis["On_Time"]
    / category_analysis["Non_Canceled"]
    * 100
)

# Average delay among delayed shipments
category_delay_avg = (
    df[df["Delivery Performance"] == "Delayed"]
    .groupby("Category Name")["Delay Gap"]
    .mean()
    .reset_index(name="Avg_Delay_When_Delayed")
)

category_analysis = category_analysis.merge(
    category_delay_avg,
    on="Category Name",
    how="left"
)

# Round numerical metrics
percentage_columns = [
    "Avg_Delay_Gap",
    "Delay_Rate_%",
    "On_Time_Rate_%",
    "Avg_Delay_When_Delayed"
]

category_analysis[percentage_columns] = (
    category_analysis[percentage_columns].round(2)
)

# Sort by delay rate
category_analysis = category_analysis.sort_values(
    "Delay_Rate_%",
    ascending=False
)

print("==========================================")
print("CATEGORY DELIVERY ANALYSIS")
print("==========================================")

display(category_analysis)
```

```python
# ==========================================
# DAY 6 - STEP 1
# LOAD ANALYSIS DATA
# ==========================================

import pandas as pd
import numpy as np
import os

analysis_path = r"D:\APL_Logistics_Project\04_Analysis"

# Load main analysis dataset
analysis_file = rf"{analysis_path}\APL_Logistics_Analysis.csv"

analysis_df = pd.read_csv(
    analysis_file,
    encoding="utf-8-sig"
)

print("==========================================")
print("DAY 6 - DATA LOADED")
print("==========================================")

print("Shape:", analysis_df.shape)

print("\nColumns:")
print(analysis_df.columns.tolist())

print("\nFirst 5 rows:")
display(analysis_df.head())
```

```python
# ==========================================
# DAY 6 - STEP 2
# MASTER KPI SUMMARY
# ==========================================

total_shipments = len(analysis_df)

delayed_shipments = (
    analysis_df["Delivery Performance"] == "Delayed"
).sum()

on_time_shipments = (
    analysis_df["Delivery Performance"] == "On-Time"
).sum()

early_shipments = (
    analysis_df["Delivery Performance"] == "Early"
).sum()

canceled_shipments = (
    analysis_df["Delivery Performance"] == "Canceled"
).sum()

non_canceled = total_shipments - canceled_shipments


# ------------------------------------------
# KPI CALCULATIONS
# ------------------------------------------

delay_rate = (
    delayed_shipments / non_canceled * 100
)

on_time_rate = (
    on_time_shipments / non_canceled * 100
)

early_rate = (
    early_shipments / non_canceled * 100
)

cancellation_rate = (
    canceled_shipments / total_shipments * 100
)

average_delay_gap = (
    analysis_df["Delay Gap"].mean()
)

delayed_only = analysis_df[
    analysis_df["Delivery Performance"] == "Delayed"
]

average_delay_when_delayed = (
    delayed_only["Delay Gap"].mean()
)

late_delivery_risk = (
    analysis_df["Late_delivery_risk"].mean() * 100
)


# ------------------------------------------
# CREATE KPI TABLE
# ------------------------------------------

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Shipments",
        "Delayed Shipments",
        "On-Time Shipments",
        "Early Shipments",
        "Canceled Shipments",
        "Delay Rate (%)",
        "On-Time Rate (%)",
        "Early Rate (%)",
        "Cancellation Rate (%)",
        "Average Delay Gap (Days)",
        "Average Delay When Delayed (Days)",
        "Late Delivery Risk (%)"
    ],

    "Value": [
        total_shipments,
        delayed_shipments,
        on_time_shipments,
        early_shipments,
        canceled_shipments,
        round(delay_rate, 2),
        round(on_time_rate, 2),
        round(early_rate, 2),
        round(cancellation_rate, 2),
        round(average_delay_gap, 2),
        round(average_delay_when_delayed, 2),
        round(late_delivery_risk, 2)
    ]
})


# ------------------------------------------
# DISPLAY
# ------------------------------------------

print("==========================================")
print("DAY 6 - MASTER KPI SUMMARY")
print("==========================================")

display(kpi_summary)
```

```python
# ==========================================
# DAY 6 - STEP 3
# DELIVERY PERFORMANCE VISUALIZATION DATA
# ==========================================

delivery_performance_viz = (
    analysis_df["Delivery Performance"]
    .value_counts()
    .reindex(
        ["Delayed", "On-Time", "Early", "Canceled"],
        fill_value=0
    )
    .reset_index()
)

delivery_performance_viz.columns = [
    "Delivery Performance",
    "Shipments"
]

# Calculate percentage of all shipments
delivery_performance_viz["Percentage_%"] = (
    delivery_performance_viz["Shipments"]
    / len(analysis_df)
    * 100
)

delivery_performance_viz["Percentage_%"] = (
    delivery_performance_viz["Percentage_%"]
    .round(2)
)

print("==========================================")
print("DELIVERY PERFORMANCE VISUALIZATION DATA")
print("==========================================")

display(delivery_performance_viz)
```

```python
# ==========================================
# DAY 6 - STEP 4
# DELAY GAP DISTRIBUTION
# ==========================================

delay_gap_viz = (
    analysis_df[
        analysis_df["Delay Gap"] > 0
    ]["Delay Gap"]
    .value_counts()
    .sort_index()
    .reset_index()
)

delay_gap_viz.columns = [
    "Delay_Gap_Days",
    "Delayed_Shipments"
]

# Percentage of all delayed shipments
total_delayed = delay_gap_viz["Delayed_Shipments"].sum()

delay_gap_viz["Share_of_Delayed_%"] = (
    delay_gap_viz["Delayed_Shipments"]
    / total_delayed
    * 100
).round(2)

# Classify severity
delay_gap_viz["Severity"] = np.select(
    [
        delay_gap_viz["Delay_Gap_Days"] == 1,
        delay_gap_viz["Delay_Gap_Days"] == 2,
        delay_gap_viz["Delay_Gap_Days"] >= 3
    ],
    [
        "Minor Delay",
        "Moderate Delay",
        "Severe Delay"
    ],
    default="Unknown"
)

print("==========================================")
print("DELAY GAP DISTRIBUTION")
print("==========================================")

display(delay_gap_viz)
```
